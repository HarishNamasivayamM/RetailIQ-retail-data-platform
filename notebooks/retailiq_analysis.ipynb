# RetailIQ analysis notebook

This notebook explains the verified synthetic baseline used by the RetailIQ data platform. It reads the same metrics artifact used by the static report and Streamlit dashboard.

```python
from pathlib import Path
import json
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path.cwd()
if not (ROOT / 'artifacts' / 'sample_metrics.json').exists():
    ROOT = ROOT.parent
metrics = json.loads((ROOT / 'artifacts' / 'sample_metrics.json').read_text(encoding='utf-8'))
plt.style.use('seaborn-v0_8-whitegrid')
```

## Headline KPIs

```python
kpis = metrics['kpis']
pd.DataFrame({
    'Metric': ['Revenue', 'Profit', 'Margin', 'Transactions', 'Average order value', 'Units sold', 'Revenue growth'],
    'Value': [
        f"${kpis['total_revenue']:,.2f}",
        f"${kpis['total_profit']:,.2f}",
        f"{kpis['profit_margin_pct']:.2f}%",
        f"{kpis['total_transactions']:,}",
        f"${kpis['average_order_value']:,.2f}",
        f"{kpis['units_sold']:,}",
        f"{kpis['revenue_growth_pct']:.2f}%",
    ]
})
```

## Monthly revenue

```python
monthly = pd.DataFrame(metrics['monthly_revenue'])
monthly.plot(x='month', y='revenue', figsize=(14, 5), marker='o', legend=False, color='#1f6feb')
plt.title('Monthly revenue')
plt.ylabel('Revenue ($)')
plt.xlabel('Month')
plt.xticks(rotation=45)
plt.tight_layout()
```

## Category and store performance

```python
categories = pd.DataFrame(metrics['category_performance']).sort_values('revenue')
fig, axes = plt.subplots(1, 2, figsize=(15, 5))
categories.plot.barh(x='category', y='revenue', ax=axes[0], legend=False, color='#1f6feb')
axes[0].set_title('Revenue by category')
axes[0].set_xlabel('Revenue ($)')
stores = pd.DataFrame(metrics['store_performance']).sort_values('revenue')
stores.plot.barh(x='store_location', y='revenue', ax=axes[1], legend=False, color='#16803c')
axes[1].set_title('Revenue by store')
axes[1].set_xlabel('Revenue ($)')
plt.tight_layout()
```

## Top products and interpretation

Electronics dominates revenue, while Home and Apparel show stronger margins. Laptop Pro 14 is the largest individual product by revenue. These are synthetic-data findings and should be treated as a demonstration of the analytical workflow, not as real business performance.
